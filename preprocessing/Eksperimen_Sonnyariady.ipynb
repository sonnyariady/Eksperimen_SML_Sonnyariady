Notebook ini digunakan untuk proses EDA dan preprocessing dataset.

```python
from pathlib import Path
import random
import matplotlib.pyplot as plt
from PIL import Image

# Cari lokasi notebook / project
try:
    BASE_DIR = Path(__file__).resolve().parent
except:
    # fallback untuk Jupyter Notebook
    BASE_DIR = Path.cwd()

print("BASE_DIR:", BASE_DIR)

# Cari root project otomatis
possible_roots = [
    BASE_DIR,
    BASE_DIR.parent,
    BASE_DIR.parent.parent,
]

project_root = None

for root in possible_roots:
    dataset_candidate = root / "namadataset_preprocessing"

    if dataset_candidate.exists():
        project_root = root
        break

if project_root is None:
    raise FileNotFoundError(
        "Folder namadataset_preprocessing tidak ditemukan."
    )

print("PROJECT ROOT:", project_root)

dataset_path = project_root / "namadataset_preprocessing"

print("DATASET PATH:", dataset_path)

classes = [d.name for d in dataset_path.iterdir() if d.is_dir()]

print("Jumlah class:", len(classes))
print(classes[:10])
```

```python
class_counts = {}

for class_name in classes:
    class_dir = dataset_path / class_name
    class_counts[class_name] = len(list(class_dir.glob('*')))

class_counts
```

```python
plt.figure(figsize=(12,6))
plt.bar(class_counts.keys(), class_counts.values())
plt.xticks(rotation=90)
plt.title('Distribusi Dataset')
plt.tight_layout()
plt.show()
```

```python
sample_classes = classes[:6]

fig, axes = plt.subplots(2, 3, figsize=(12,8))

for ax, class_name in zip(axes.flatten(), sample_classes):
    class_dir = dataset_path / class_name
    image_path = random.choice(list(class_dir.glob('*')))
    img = Image.open(image_path)
    ax.imshow(img)
    ax.set_title(class_name)
    ax.axis('off')

plt.tight_layout()
plt.show()
```
